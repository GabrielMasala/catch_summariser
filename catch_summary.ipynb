```python
import pandas as pd
landings = pd.read_csv("landings.csv")
```

```python
pd.to_numeric(landings["weight_kg"], errors="coerce")
```

```python
landings["weight_kg"] = pd.to_numeric(landings["weight_kg"], errors="coerce")
```

```python
landings["weight_kg"].isna().sum()
```

```python
landings["weight_kg"].notna().sum()
```

```python
landings["weight_kg"].isna().sum() + landings["weight_kg"].notna().sum() == len(landings)
```

```python
species_totals = landings.groupby("species")["weight_kg"].sum(min_count=1)
```

```python
species_totals
```

```python
sites_totals = landings.groupby("landing_site")["weight_kg"].sum(min_count=1)
```

```python
sites_totals
```

```python
species_totals.sum()
```

```python
sites_totals.sum()
```

```python
species_totals.sum() == sites_totals.sum()
```
